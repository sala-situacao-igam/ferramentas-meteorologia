```python
import os
import zipfile
import threading
import time
import shutil
from google.colab import output

# ==========================================
# TEMPO SEVERO WEB - V0.17
# AJUSTE FINO DO LAYOUT
# ==========================================

zip_path = "/content/tempo_severo_web_v0_17.zip"
destino = "/content/tempo_severo_web_v0_17_ajustado"

# Limpa teste anterior
if os.path.exists(destino):
    shutil.rmtree(destino)

os.makedirs(destino, exist_ok=True)

# Extrai o ZIP
with zipfile.ZipFile(zip_path, "r") as zip_ref:
    zip_ref.extractall(destino)

print("✅ ZIP extraído com sucesso!")

# Procura o index.html
index_path = None

for raiz, pastas, arquivos in os.walk(destino):
    if "index.html" in arquivos:
        index_path = os.path.join(raiz, "index.html")
        break

if index_path is None:
    raise FileNotFoundError("❌ Não encontrei o index.html.")

print("✅ index.html encontrado:")
print(index_path)

# ==========================================
# AJUSTES FINOS
# ==========================================

with open(index_path, "r", encoding="utf-8") as f:
    html = f.read()

# 1. Legenda - mudar a posição
html = html.replace(
    "#printLegend{position:absolute;right:24px;bottom:92px;",
    "#printLegend{position:absolute;right:66px;bottom:92px;"
)

# 2. Mudar posicao da Escala + seta Norte
html = html.replace(
    "const x=38,y=ctx.canvas.height-62,w=80;",
    "const x=58,y=ctx.canvas.height-62,w=80;"
)

# 3. Mudar a posição do Texto de validade
html = html.replace(
    "#printValidity{position:absolute;right:18px;",
    "#printValidity{position:absolute;right:38px;"
)

# Salva o index ajustado
with open(index_path, "w", encoding="utf-8") as f:
    f.write(html)


# ==========================================
# INICIA O SISTEMA
# ==========================================

pasta_sistema = os.path.dirname(index_path)
os.chdir(pasta_sistema)

PORTA = 8018

def iniciar_servidor():
    os.system(f"python3 -m http.server {PORTA}")

thread = threading.Thread(target=iniciar_servidor)
thread.daemon = True
thread.start()

time.sleep(4)

print("✅ Servidor iniciado!")

url = output.eval_js(
    f"google.colab.kernel.proxyPort({PORTA})"
)

print("\n===================================")
print("🌐 ABRA O SISTEMA NESTE LINK:")
print(url)
print("===================================")
```
