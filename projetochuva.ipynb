```python
import os
import zipfile
import threading
import time
from google.colab import output

# ==========================================
# CHUVA + TENDÊNCIA WEB - V0.3
# ==========================================

zip_path = "/content/chuva_tendencia_web_v0_3.zip"
destino = "/content/chuva_tendencia_web_v0_3"

if os.path.exists(destino):
    import shutil
    shutil.rmtree(destino)

os.makedirs(destino, exist_ok=True)

with zipfile.ZipFile(zip_path, "r") as zip_ref:
    zip_ref.extractall(destino)

print("✅ ZIP extraído com sucesso!")

index_encontrado = None

for raiz, pastas, arquivos in os.walk(destino):
    if "index.html" in arquivos:
        index_encontrado = raiz
        break

if index_encontrado is None:
    raise FileNotFoundError("❌ Não encontrei o index.html.")

print("✅ Sistema encontrado em:")
print(index_encontrado)

os.chdir(index_encontrado)

PORTA = 8023

def iniciar_servidor():
    os.system(f"python3 -m http.server {PORTA}")

thread = threading.Thread(target=iniciar_servidor)
thread.daemon = True
thread.start()

time.sleep(4)

print("✅ Servidor iniciado!")

url = output.eval_js(
    f"google.colab.kernel.proxyPort({PORTA})"
)

print("\n===================================")
print("🌐 ABRA O SISTEMA NESTE LINK:")
print(url)
print("===================================")
```
